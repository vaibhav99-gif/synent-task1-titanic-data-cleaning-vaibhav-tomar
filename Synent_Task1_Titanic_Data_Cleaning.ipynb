# Task 1: Data Cleaning and Preprocessing

**Objective:** The objective of this project is to clean and prepare the Titanic dataset for further analysis.

Data cleaning is an important step in Data Science because raw datasets may contain missing values, duplicate records, inconsistent column names, and incorrect data types. In this project, I will perform basic preprocessing using Python and Pandas.

Pandas is used to load, manipulate, and clean tabular data. NumPy provides numerical computing functionality. These libraries help perform the data preprocessing steps efficiently.

```python
import pandas as pd
import numpy as np
```

The Titanic dataset is loaded into a Pandas DataFrame named `df`. It contains passenger information such as age, gender, ticket class, fare, and survival status. The `head()` function displays the first five rows to help us understand the structure of the dataset.

```python
import seaborn as sns

df = sns.load_dataset("titanic")

df.head()
```

This step examines the dataset before cleaning. The `shape` attribute returns the number of rows and columns. The column list shows the available features, `isnull().sum()` counts missing values in each column, and `duplicated().sum()` counts duplicate rows.

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

The `drop_duplicates()` function removes rows that are identical across all columns. This helps prevent repeated records from affecting subsequent analysis. The updated shape shows the dataset size after duplicate removal.

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

Missing values can reduce data quality and affect analysis. Missing ages are replaced with the median age, which is less sensitive to extreme values than the mean. Missing embarkation categories are replaced with their most frequent values, calculated using the mode.

```python
print(df.isnull().sum())
```

Missing values can reduce data quality and affect analysis. Missing ages are replaced with the median age, which is less sensitive to extreme values than the mean. Missing embarkation categories are replaced with their most frequent values, calculated using the mode.

```python
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
```

```python
print(df.isnull().sum())
```

This step ensures that numerical columns are stored as numeric values and categorical columns are represented as categories. The `errors="coerce"` option converts invalid numeric values to missing values, which should be checked during the final quality review.

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["fare"] = pd.to_numeric(df["fare"], errors="coerce")
```

```python
df["sex"] = df["sex"].astype("category")
df["class"] = df["class"].astype("category")
```

This step ensures that numerical columns are stored as numeric values and categorical columns are represented as categories. The `errors="coerce"` option converts invalid numeric values to missing values, which should be checked during the final quality review.

```python
df = df.rename(columns={
    "survived": "Survived",
    "pclass": "Passenger_Class",
    "sex": "Gender",
    "age": "Age",
    "fare": "Fare"
})

df.head()
```

The `rename()` function changes selected column names to more readable labels. Clear column names make the dataset easier to understand and work with. Only the specified columns are renamed; the remaining column names stay unchanged.

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

The final validation checks the dataset's dimensions, remaining missing values, duplicate rows, and data types. This helps confirm whether the preprocessing steps were effective and identifies any remaining data-quality issues that may need attention.

```python
df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

The `to_csv()` function exports the cleaned DataFrame to a CSV file. Setting `index=False` prevents Pandas from adding an extra index column.
